````python
import re
import time
import traceback
from pathlib import Path
from urllib.request import urlopen

import pandas as pd
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from tqdm import tqdm
from webdriver_manager.chrome import ChromeDriverManager
````

# 東進のデータ
## 各情報取得用のURLを抽出する

```Python
https://www.toshin-hensachi.com/?line=2&university_type={ここに大学タイプ}&page={ここにページ}&sort=deviation&direction=DESC
```

````python
data_list = []
for index,university_type in tqdm(enumerate([11,12,13,21,22,23])):
    for page in range(1,[12,1,13,15,1,9][index] +1):
        #^ __init__
        score =0
        url = f"https://www.toshin-hensachi.com/?line=2&university_type={university_type}&page={page}&sort=deviation&direction=DESC"
        print(url)
        a_url_list = [] #~大学詳細ページのURLを格納するリスト
        html = urlopen(url).read()  #^ スクレイピング
        time.sleep(1)  # 1秒待つ
        soup = BeautifulSoup(html, "html.parser")
        table = soup.find_all("table",class_="table-ranking")
        t_body= table[0].find("tbody")
        tr_list = t_body.find_all("tr")
        for tr in tr_list:
            try:
                #^ 偏差値
                td_score = tr.find("td",class_="td-score")
                if td_score is not None:
                    score = td_score.text.strip()
                    td_score = score
                else:
                    td_score = score
                #^ 大学情報
                td_info = tr.find("td", class_="td-info")
                try:
                    school_name_data = td_info.find("p", class_="school-name")
                except:
                    continue
                school_link = school_name_data.find("a")
                if school_name_data:
                    a_tag = school_name_data.find("a",href=True)
                    if a_tag is not None:
                        univ_url = a_tag.get("href")
                    school_name_tag = school_name_data.find("span", class_="school-name-university")
                    if school_name_tag:
                        school_name = school_name_tag.text.strip()
                        department = school_name_data.text.replace(school_name, "").strip()
                    else:
                        department = "学部情報なし"
                else:
                    school_name = "学校情報なし"
                    department = "学部情報なし"
                magnification_urls = td_info.find_all("a",class_="link-pasttest")
                for a in magnification_urls:
                    text = a.get_text(strip=True)
                    if text=="倍率を見る":
                        #^ 倍率を見る
                        magnification_url = a.get("href")
                experience_urls = td_info.find_all("a",class_="link-experience")
                for a in experience_urls:
                    if a.text.strip()=="合格体験記を見る":
                        #^ 合格体験記
                        SuccessfulExperience = a.get("href")
                    if a.text.strip()=="入試日程を見る":
                        #^ 入試日程
                        EntranceExamSchedule = a.get("href")
                dict= {"大学名":school_name,
                    "学部":department,
                    "大学詳細ページ":univ_url,
                    "偏差値": td_score,
                    "倍率情報": magnification_url,
                    "合格体験記": SuccessfulExperience,
                    "入試日程": EntranceExamSchedule,
                    }
                data_list.append(dict)
            except Exception as e:
                print("Error",tr)
                display(traceback.print_exc())  # どの行でエラーが発生したか表示
````

````python
# データをデータフレームとして格納し、csv形式で保存する
df = pd.DataFrame(data_list)
df.to_csv("toshin_university_data.csv", index=True, encoding="utf-8-sig")
````
